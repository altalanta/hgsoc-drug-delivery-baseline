# PK sensitivity scratch
Quick check of PK shifts on response probability.

```python
from hgsoc.simulate import simulate_cohort
from hgsoc.features import prepare_features
from sklearn.linear_model import LogisticRegression
import numpy as np
df = simulate_cohort(300, seed=2024)
bundle = prepare_features(df)
model = LogisticRegression(max_iter=500).fit(bundle.X_train, bundle.y_train)
probs = model.predict_proba(bundle.X_valid)[:, 1]
probs.mean()
```
